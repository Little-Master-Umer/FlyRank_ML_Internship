## 2. Fields: feature / label / context / excluded

### Field Categorization

Based on the typical structure of search intelligence data and the prediction goal, here's a categorization of the fields:

*   **Features:** These are the independent variables used to make predictions.
    *   `query_text`: The user's search query (categorical/textual).
    *   `search_engine`: The search engine used (categorical).
    *   `device_type`: The device from which the search originated (e.g., mobile, desktop, tablet) (categorical).
    *   `user_id`: Anonymized user identifier (categorical).
    *   `time_of_day`: Derived from `timestamp`, representing the hour or time bucket of the search (numerical/categorical).
    *   `day_of_week`: Derived from `timestamp`, representing the day of the week (numerical/categorical).
    *   `num_results_displayed`: The number of search results shown (numerical).
    *   `result_rank`: The position of a specific result in the search listing (numerical).
    *   `query_length`: The number of characters or words in `query_text` (numerical).
    *   `result_title`: Title of the search result (textual).
    *   `result_url`: URL of the search result (textual).

*   **Labels:** These are the dependent variables we aim to predict.
    *   `result_clicked`: Binary indicator (0 or 1) if a search result was clicked.
    *   `click_rank`: The rank of the clicked result (numerical) – if a click occurred.

*   **Context:** These fields provide additional information but may not be directly used as features in a simple model, or are used to derive features.
    *   `timestamp`: Exact time of the search event (datetime).
    *   `session_id`: Identifier for a user session (categorical).
    *   `page_number`: The search results page number (numerical).
    *   `query_vertical`: The type of search (e.g., web, image, video) (categorical).

*   **Excluded:** These fields are explicitly removed due to privacy, high cardinality, or irrelevance.
    *   `raw_ip_address`: (Privacy) Directly identifiable information.
    *   `user_agent_string`: (High Cardinality/Privacy) Often contains browser and OS details that can be very specific, making it difficult to generalize and potentially linking to specific users.
    *   `exact_location_coordinates`: (Privacy) Too granular and directly identifiable.
